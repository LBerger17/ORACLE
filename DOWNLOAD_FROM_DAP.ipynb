For AWAKEN - One file is stored every 15 seconds with all these variables				

Code	Data Description	Format	Range	Unit
- 1	Rain Intensity	Float	0.0 to 9999.999	milli-meter per hour
- 2	Rain amount accumulated	Float	0.00 to 300.00	milli-meter

- 19	Date/time measuring start		DD.MM.YYYY_hh:mm:ss
- 20	Sensor time		hh:mm:ss
- 21	Sensor date		DD.MM.YYYY

- 24	Rain amount absolute	Float	0.000 to 999.999	milli-meter

- 30	Rain intensity (16 bit) max. 30.000 mm/h	Float	0.000 to 30.000	milli-meter per hour
- 31	Rain intensity (16 bit) max. 1200.0 mm/h	Float	0.0 to 1200.0	milli-meter per hour
- 32	Rain amount accmulated (16 bit)	Float	0.00 to 300.00	milli-meter
