```python
import os

# Open first CDF file from bnfldM1.b1
bnfld_path = "/people/gold396/ORACLE/ARM/ftp.archive.arm.gov/goldbergerl2/264931/bnfldM1.b1/bnfldM1.b1.20260320.000000.cdf"
bnfld_quants_path = "/people/gold396/ORACLE/ARM/ftp.archive.arm.gov/goldbergerl2/264931/bnfldquantsM1.c1/bnfldquantsM1.c1.20260320.000000.nc"

ds_bnfld = xr.open_dataset(bnfld_path)
ds_bnfld_quants = xr.open_dataset(bnfld_quants_path)

# Compare with ds
print("=== bnfldM1.b1 vs ds ===")
bnfld_vars = set(ds_bnfld.data_vars)
match_bnfld = bnfld_vars == ds_vars
print(f"Same variable names: {match_bnfld}")
print(f"Number of vars in bnfldM1.b1: {len(bnfld_vars)}, in ds: {len(ds_vars)}")
if not match_bnfld:
    only_in_bnfld = bnfld_vars - ds_vars
    only_in_ds_bnfld = ds_vars - bnfld_vars
    print(f"Only in bnfldM1.b1 ({len(only_in_bnfld)}): {sorted(only_in_bnfld)[:10]}")
    print(f"Only in ds ({len(only_in_ds_bnfld)}): {sorted(only_in_ds_bnfld)[:10]}")

print("\n=== bnfldquantsM1.c1 vs ds ===")
bnfld_quants_vars = set(ds_bnfld_quants.data_vars)
match_quants = bnfld_quants_vars == ds_vars
print(f"Same variable names: {match_quants}")
print(f"Number of vars in bnfldquantsM1.c1: {len(bnfld_quants_vars)}, in ds: {len(ds_vars)}")
if not match_quants:
    only_in_quants = bnfld_quants_vars - ds_vars
    only_in_ds_quants = ds_vars - bnfld_quants_vars
    print(f"Only in bnfldquantsM1.c1 ({len(only_in_quants)}): {sorted(only_in_quants)[:10]}")
    print(f"Only in ds ({len(only_in_ds_quants)}): {sorted(only_in_ds_quants)[:10]}")
```
